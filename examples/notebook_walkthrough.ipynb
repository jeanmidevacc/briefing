# Building a report interactively

This is the notebook workflow briefing is built for: each cell below creates one block and ends with it, so the notebook previews it immediately (each preview is isolated in its own frame), the same way you'd build a report while exploring data. The last cell combines every block created above into a single `Briefing` and saves it as one self-contained HTML file.

```python
import numpy as np
import pandas as pd
import briefing as bf

blocks = []  # every block built below gets appended here

rng = np.random.default_rng(7)
df = pd.DataFrame({
    "region": rng.choice(["North", "South", "East", "West"], size=200),
    "revenue": rng.normal(12_000, 3_000, size=200).round(2),
    "margin_pct": rng.normal(0.31, 0.05, size=200).round(3),
})
```

## 1. Start with a headline

```python
intro = bf.Text("# Sales Analysis, Q1 2024")
blocks.append(intro)
intro  # a block previews itself as the last expression of a cell
```

## 2. Add the KPIs

```python
kpis = bf.Group(
    bf.BigNumber("Revenue", f"${df['revenue'].sum():,.0f}", change="+12%", is_upward_change=True),
    bf.BigNumber("Avg Margin", f"{df['margin_pct'].mean():.1%}", change="-0.8%", is_upward_change=False),
    columns=2,
)
blocks.append(kpis)
kpis  # a block previews itself as the last expression of a cell
```

## 3. Add a chart

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3.5))
df.groupby("region")["revenue"].mean().plot.bar(ax=ax, color="#4F46E5")
ax.set_ylabel("Avg revenue")
ax.set_title("Revenue by region")

chart = bf.Plot(fig, caption="Average revenue per region")
blocks.append(chart)
chart  # a block previews itself as the last expression of a cell
```

## 4. Add the raw data

```python
table = bf.DataTable(df, caption="All transactions")
blocks.append(table)
table  # a block previews itself as the last expression of a cell
```

## 5. Add a column profile

```python
profile = bf.lab.DataProfile(df)
blocks.append(profile)
profile  # a block previews itself as the last expression of a cell
```

## 6. Put it all together

```python
report = bf.Briefing(*blocks)
bf.save(report, "sales_report.html", name="Sales Analysis, Q1 2024")
bf.preview(report, name="Sales Analysis, Q1 2024")  # with the name it is saved under
```
